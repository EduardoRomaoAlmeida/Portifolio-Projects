# U.S. Medical Insurance Costs Analysis

## Objective

Investigate which personal characteristics influence the cost of medical insurance,
using a dataset with information on 1,338 people in the United States.

## Data
The `insurance.csv` file contains the following variables:
- **age**: age of the person
- **sex**: sex (female / male)
- **bmi**: body mass index (BMI)
- **children**: number of children covered by the insurance
- **smoker**: whether the person smokes (yes / no)
- **region**: region of residence in the U.S. (northeast, northwest, southeast, southwest)
- **charges**: annual medical insurance cost, in U.S. dollars

## Questions

### Part 1: Getting to know the data
1. What are the mean, most frequent value, minimum and maximum of each numeric variable (age, BMI, children and charges)?
2. What is the most common category of each text variable (sex, smoker and region)? How are people distributed among them?

### Part 2: Predicting the cost
3. Is it possible to create a formula that estimates the insurance cost from the other variables?
4. Using this formula, how much would a person with a specific profile pay?

### Part 3: What drives the cost
5. Which three variables have the greatest impact on insurance cost?
6. How much more does a smoker pay, on average, than a non-smoker?
7. Does insurance cost increase with age and BMI?
""""""

```python
import csv

ages = []
sexes = []
bmis = []
num_children = []
smoker_status = []
regions = []
insurance_charges = []

with open('insurance.csv') as insurance_file:
    insurance_data = csv.DictReader(insurance_file)
    for row in insurance_data:
        ages.append(int(row['age']))
        sexes.append(row['sex'])
        bmis.append(float(row['bmi']))
        num_children.append(int(row['children']))
        smoker_status.append(row['smoker'])
        regions.append(row['region'])
        insurance_charges.append(float(row['charges']))
```

```python
def average(data):
    return sum(data) / len(data)

def mode(data):
    return max(set(data), key=data.count)

def summary(name, data):
    print(f"{name}:")
    print(f"  Mean: {round(average(data), 2)}")
    print(f"  Most frequent value: {mode(data)}")
    print(f"  Minimum: {min(data)}")
    print(f"  Maximum: {max(data)}\n")

def category_summary(name, values):
    total = len(values)
    print(f"{name}:")
    print(f"  Most frequent: {mode(values)}")
    for category in sorted(set(values)):
        count = values.count(category)
        percentage = count / total * 100
        print(f"  {category}: {count} of {total} ({round(percentage, 1)}%)")
    print()

category_summary("Sex", sexes)
category_summary("Smoker", smoker_status)
category_summary("Region", regions)

summary("Age", ages)
summary("BMI", bmis)
summary("Number of Children", num_children)
summary("Insurance Charges", insurance_charges)

```

#Health Insurance Dataset – Summary and Key Insights
The dataset contains 1,338 observations of health insurance beneficiaries in the United States.
- Sex: Nearly balanced distribution between males (50.5%) and females (49.5%).
- Smoking: Most beneficiaries are non-smokers (79.5%), while 20.5% are smokers.
- Region: Relatively balanced distribution across all four regions, with the Southeast having the largest share (27.2%).
- Age: Average age of 39.2 years, ranging from 18 to 64.
- BMI: Average BMI of 30.66, indicating obesity on average.
- Children: Average of 1.09 children per beneficiary, with 0 being the most common value.
- Insurance Charges: Average annual charges of $13,270.42, ranging from $1,121.87 to $63,770.43.
Key Insights
The dataset shows balanced demographic and geographic representation. The high average BMI and the proportion of smokers highlight potential health risk factors worth investigating.

Insurance charges vary significantly, suggesting a potentially right-skewed distribution with high-cost cases. Age, BMI, and smoking status are important variables to examine as potential predictors of insurance expenses.

Further analysis is needed to determine how strongly these factors are associated with insurance charges.

```python
sex_numeric = [1 if s == "male" else 0 for s in sexes]
smoker_numeric = [1 if s == "yes" else 0 for s in smoker_status]
northwest = [1 if r == "northwest" else 0 for r in regions]
southeast = [1 if r == "southeast" else 0 for r in regions]
southwest = [1 if r == "southwest" else 0 for r in regions]

X = [list(row) for row in zip(ages, sex_numeric, bmis, num_children,
                              smoker_numeric, northwest, southeast, southwest)]
y = insurance_charges

print(X[0])  # → [19, 0, 27.9, 0, 1, 0, 0, 1]

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

feature_names = ["age", "sex (male)", "bmi", "children", "smoker",
                 "northwest", "southeast", "southwest"]

print(f"Intercept: {round(model.intercept_, 2)}")
for name, coef in zip(feature_names, model.coef_):
    print(f"{name}: {round(coef, 2)}")

print(f"R²: {round(model.score(X, y), 3)}")
```


Linear Regression – Results and Key Insights

The linear regression model explains approximately 75.1% of the variation
in health insurance charges (R² = 0.751), indicating a good overall fit.

Key Insights:

- Smoking is the strongest predictor, increasing expected insurance charges
  by approximately $23,849 compared to non-smokers.

- Age: Each additional year is associated with a $257 increase in charges.

- BMI: Each additional BMI unit is associated with a $339 increase in charges.

- Children: Each additional child increases expected charges by $476.

- Sex: Males have slightly lower predicted charges (-$131) than females.

- Region: Compared to the Northeast (reference category), all other regions
  have lower predicted charges, particularly the Southeast (-$1,035).

Conclusion:

Smoking status has the largest impact on predicted insurance charges,
followed by BMI and age. The model explains a substantial proportion
of the variation in healthcare costs, although approximately 24.9%
remains unexplained.

All coefficients represent associations while holding other variables
constant. Statistical significance has not been evaluated.
